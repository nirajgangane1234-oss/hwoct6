```python
Total_student = ["Raj", "Amit", "Jay", "Payal", "Priya",
                 "Rahul", "Pavan", "Kiran", "Prachi"]
present_attendance = ("Raj", "Jay", "Payal", "Priya", "Pavan", "Kiran")
for student in Total_student:
    if present_attendance.count(student) == 0:
        print(student, "was absent today")
```

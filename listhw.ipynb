```python
# Append Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
print(l)
l.append("Shantanu")
print(l)
```

```python
# Len Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
print(len(l))
```

```python
# Insert Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
l.insert(2,"Aayush")
print(l)
```

```python
# Extend Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
l.extend(["Shantanu","Abhinay"])
print(l)
```

```python
# Remove Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
l.remove("Niraj")
print(l)
```

```python
# Pop Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
removed_list = l.pop()
print("Removed using pop:", removed_list)
print("After pop:", l)
```

```python
# Count Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
print(l.count("Niraj"))
```

```python
# Index Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
print(l.index("Hrishi"))
```

```python
# Sort Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
l.sort()
print(l)
```

```python
# Reverse Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
l.reverse()
print(l)
```

```python
# Copy Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
new_list = l.copy()
print(new_list)
```

```python
# Clear Method

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
new_list = l.copy()

print("Copied List:", new_list)

new_list.clear()

print("After clear:", new_list)
```

```python
# Membership Operation

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
print("Is Sahil present?", "Sahil" in l)
print("Is Abhinay present?", "Abhinay" in  l)
```

```python
# Not in

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
print("Is Arpan not present?", "Arpan" not in l)
```

```python
# Indexing

l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
print("First name:", l[0])
print("Last name:", l[-1])
```

```python
# Slicing


l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
print("First 3 list:", l[:3])
print("From index 2:", l[2:])
```

```python
# For loop


l = ["Niraj","Arpan","Sahil","Vedant","Hrishi"]
print("\nAll Names")
for list in l:
    print(list)
```
